# Calculating the relative density between depths of -20 m and -25 using the Robertson method

```python
import pandas as pd
from pathlib import Path
from IPython.display import display, Markdown
```

```python
csv_path_1 = Path(r"C:\AA_Thesis\VSCode\Horizontal_stress\Ahmadi_golestani\CPT1-2\CPT1-2_bottom_sand_Qt_Dr.csv")
csv_path_2 = Path(r"C:\AA_Thesis\VSCode\Horizontal_stress\Ahmadi_golestani\CPT1-4\CPT1-4_bottom_sand_Qt_Dr.csv")
csv_path_3 = Path(r"C:\AA_Thesis\VSCode\Horizontal_stress\Ahmadi_golestani\CPT2-2\CPT2-2_bottom_sand_Qt_Dr.csv")
csv_path_4 = Path(r"C:\AA_Thesis\VSCode\Horizontal_stress\Ahmadi_golestani\CPT2-4\CPT2-4_bottom_sand_Qt_Dr.csv")

# =============================================================================
# LOAD CSV FILES
# =============================================================================

df_1_2 = pd.read_csv(csv_path_1)
df_1_4 = pd.read_csv(csv_path_2)
df_2_2 = pd.read_csv(csv_path_3)
df_2_4 = pd.read_csv(csv_path_4)

display(df_1_2.head())
```

```python
# =============================================================================
# AVERAGE Dr ABOVE, BETWEEN, AND BELOW -20.0 TO -25.0
# =============================================================================

dfs = {
    "df_1_2": df_1_2,
    "df_1_4": df_1_4,
    "df_2_2": df_2_2,
    "df_2_4": df_2_4,
}

results_b = []   # between -20 and -25
results_u = []   # under / deeper than -25
results_a = []   # above / shallower than -20

for name, df in dfs.items():
    depth_col = df.columns[1]      # second depth giving true vertical depth
    dr_col = df.columns[-1]        # last column = Dr

    df_temp = df.copy()

    df_temp[depth_col] = pd.to_numeric(df_temp[depth_col], errors="coerce")
    df_temp[dr_col] = pd.to_numeric(df_temp[dr_col], errors="coerce")

    # Between -20 and -25 m NAP
    between = df_temp[
        (df_temp[depth_col] >= 20.0)
        & (df_temp[depth_col] <= 22.5)
    ].copy()

    # Under / deeper than -25 m NAP
    under = df_temp[
        df_temp[depth_col] < 22.5
    ].copy()

    # Above / shallower than -20 m NAP
    above = df_temp[
        df_temp[depth_col] > 20.0
    ].copy()

    results_b.append({
        "csv": name,
        # "depth_column": depth_col,
        # "dr_column": dr_col,
        "n_values": len(between),
        # "selected_z_values": between[depth_col].tolist(),
        # "selected_Dr_values": between[dr_col].tolist(),
        "mean_Dr": between[dr_col].mean(),
    })

    results_u.append({
        "csv": name,
        # "depth_column": depth_col,
        # "dr_column": dr_col,
        "n_values": len(under),
        # "selected_z_values": under[depth_col].tolist(),
        # "selected_Dr_values": under[dr_col].tolist(),
        "mean_Dr": under[dr_col].mean(),
    })

    results_a.append({
        "csv": name,
        # "depth_column": depth_col,
        # "dr_column": dr_col,
        "n_values": len(above),
        # "selected_z_values": above[depth_col].tolist(),
        # "selected_Dr_values": above[dr_col].tolist(),
        "mean_Dr": above[dr_col].mean(),
    })

mean_dr_df_b = pd.DataFrame(results_b)
mean_dr_df_u = pd.DataFrame(results_u)
mean_dr_df_a = pd.DataFrame(results_a)

display(Markdown("### Average Dr above -20.0 m NAP"))
display(mean_dr_df_a)

display(Markdown("### Average Dr between -20.0 and -22.5 m NAP"))
display(mean_dr_df_b)

display(Markdown("### Average Dr below -22.5 m NAP"))
display(mean_dr_df_u)
```
